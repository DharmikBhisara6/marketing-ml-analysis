# RQ7: Final Conclusion

This study analyzed marketing and product performance data using multiple machine learning models to predict customer revenue levels.

The baseline models showed moderate performance, suggesting that the dataset contains useful patterns but is not easily separable. More advanced models, particularly Random Forest, achieved better results, highlighting the effectiveness of ensemble methods in capturing complex relationships.

Feature scaling had a noticeable impact on Logistic Regression, improving performance by standardizing the range of input variables.

Feature importance analysis identified Bundle Price, Budget, and ROI as key factors influencing customer revenue. This indicates that pricing strategy and marketing efficiency play a critical role in business outcomes.

Hyperparameter tuning further improved model performance, demonstrating that carefully optimized models can achieve better predictive accuracy than default configurations.

Evaluation using the confusion matrix showed that the model performs well in identifying high-value customers but has some limitations in correctly classifying lower-value cases.

Overall, the results demonstrate that machine learning can provide valuable insights into customer behavior and support data-driven decision-making in marketing and product strategy.
